**Step 2:** Let's begin by creating a chart that displays the average number of crashes per hour of the day. This will help us understand whether additional factors are contributing to crashes - i.e. rush hour, school dismissal time, night fall, etc.

````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Load the dataset
local_path = "/Users/dreuxx/Downloads/Motor_Vehicle_Collisions_-_Crashes.csv"
data = pd.read_csv(local_path)

# Convert 'CRASH DATE' and 'CRASH TIME' to datetime
data['CRASH DATE'] = pd.to_datetime(data['CRASH DATE'])
data['CRASH TIME'] = pd.to_datetime(data['CRASH TIME'], format='%H:%M')

# Time of Day Analysis
data['Hour of Day'] = data['CRASH TIME'].dt.hour

# Group by 'Hour of Day' and calculate the average number of crashes per hour
average_crashes_per_hour = data.groupby('Hour of Day').size() / data['Hour of Day'].nunique()

# Plot the average number of crashes
plt.figure(figsize=(12, 6))
sns.barplot(x=average_crashes_per_hour.index, y=average_crashes_per_hour.values)
plt.title('Average Number of crashes per Hour of Day')
plt.xlabel('Hour of Day')
plt.ylabel('Average Number of crashes')
plt.xticks(range(0, 24))
plt.show()
````

**Step 4**: Apply time series decomposition to review trends, seasonality, and residuals. New to time series analysis? Review the [Time Series Flashcard video series](https://youtube.com/playlist?list=PLNs9ZO9jGtUAqd0CNKySksPJJaOqYPMvk&feature=shared) here to learn about trends, components, and further analysis!

````python
%pip install statsmodels

import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.seasonal import seasonal_decompose

# Count the number of crashes per day, group by CRASH DATE
daily_crashes = data.groupby('CRASH DATE').size()

# Set plot style
sns.set(style="darkgrid")

# Plot the daily crashes time series
plt.figure(figsize=(15, 6))
plt.plot(daily_crashes.index, daily_crashes.values, label='Daily crashes')
plt.title('Daily Motor Vehicle Collisions in NYC')
plt.xlabel('Date')
plt.ylabel('Number of Crashes')
plt.legend()
plt.show()

# Decompose the time series
decomposition = seasonal_decompose(daily_crashes, model='additive', period=365)

# Plot the decomposed components
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(15, 12))
decomposition.trend.plot(ax=ax1)
ax1.set_title('Trend')
decomposition.seasonal.plot(ax=ax2)
ax2.set_title('Seasonality')
decomposition.resid.plot(ax=ax3)
ax3.set_title('Residuals')
plt.tight_layout()
plt.show()
````

The visualizations above provide valuable insights into the time series of daily motor vehicle collisions in New York City:

1. Time Series Plot: This shows the number of daily crashes over time. You might observe long-term trends, seasonal patterns, or significant outliers.

2. Decomposed Components:
  
    2.1 Trend: This graph shows the long-term trend in the data, which can indicate whether crashes are increasing, decreasing, or stable over time.

    2.2 Seasonality: This reveals any regular patterns that repeat over a specific period, such as yearly. It helps identify times of the year with higher or lower crash frequencies.

    2.3 Residuals: These are the irregular components that cannot be attributed to the trend or seasonality. They might include random or unpredictable fluctuations.

````python
#TODO: Continue building your heatmap
# Sample a subset of the data for visualization
sample_data_severity = data_geo.sample(n=1000, random_state=42)

# Create a base map
m_severity = folium.Map(location=[40.730610, -73.935242], zoom_start=10)

# Add crashes to the map with color coding and shape coding based on severity
for index, row in sample_data_severity.iterrows():
    if row['NUMBER OF PERSONS KILLED'] > 0:
        color = "red"  # Fatalities

        folium.features.RegularPolygonMarker(
          location=[row['LATITUDE'], row['LONGITUDE']],
          number_of_sides=3,
          radius=5,
          gradient = False,
          color=color,
          fill=True,
          fill_color=color
        ).add_to(m_severity)


    elif row['NUMBER OF PERSONS INJURED'] > 0:
        color = "orenge"  # Injuries
        folium.CircleMarker(
          location=[row['LATITUDE'], row['LONGITUDE']],
          radius=5,
          color=color,
          fill=True,
          fill_color=color
       ).add_to(m_severity)
    else:
        color = "green"  # No injuries or fatalities
        folium.features.RegularPolygonMarker(
          location=[row['LATITUDE'], row['LONGITUDE']],
          number_of_sides=4,
          radius=5,
          gradient = False,
          color=color,
          fill=True,
          fill_color=color
        ).add_to(m_severity)


m_severity.save("severity.html")
````

````python
%pip install pandas geopy numpy matplotlib scipy

import pandas as pd
from geopy.distance import geodesic
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial import distance

# Convert coordinates to NumPy arrays
bus_coordinates = np.array(bus_en_cleaned[['bus_latitude', 'bus_longitude']].values)
subway_coordinates = np.array(subway_en_cleaned[['subway_latitude', 'subway_longitude']].values)

# Function to calculate distances for all accidents at once using scipy
def calculate_distances(accident_locations, coordinates):
    # Access latitude and longitude using .values to get a NumPy array
    accident_coords = accident_locations[['acc_latitude', 'acc_longitude']].values
    # Calculate distances using scipy's cdist
    distances_matrix = distance.cdist(accident_coords, coordinates, 'euclidean')
    # Get the minimum distance for each accident
    min_distances = distances_matrix.min(axis=1)
    return min_distances

# Calculate distances to bus and subway stops
bus_distances = calculate_distances(data_Acc_cleaned, bus_coordinates)
subway_distances = calculate_distances(data_Acc_cleaned, subway_coordinates)

# Add distance columns to dataframe
data_Acc_cleaned['distance_to_bus'] = bus_distances
data_Acc_cleaned['distance_to_subway'] = subway_distances

# Visualización de distancias
plt.hist(data_Acc_cleaned['distance_to_subway'], bins=30, alpha=0.7, label='Subway Distance')
plt.hist(data_Acc_cleaned['distance_to_bus'], bins=30, alpha=0.7, label='Bus Distance')
plt.title('Distribución de Accidentes según Proximidad a Transporte')
plt.xlabel('Distancia (metros)')
plt.ylabel('Número de Accidentes')
plt.legend()
plt.show()
````

````python
from scipy.spatial import cKDTree

subway_Lo = subway_en[['Station Latitude', 'Station Longitude']].values
subway_tree = cKDTree(subway_Lo)

bus_Lo = bus_en[['Latitude', 'Longitude']].values
bus_tree = cKDTree(bus_Lo)

Acc_Lo = data_Acc[['LATITUDE', 'LONGITUDE']].values

subway_dist, subway_idx = subway_tree.query(Acc_Lo)
bus_dist, bus_idx = bus_tree.query(Acc_Lo)

data_Acc['Subway_Distance'] = subway_dist
data_Acc['Bus_Distance'] = bus_dist
data_Acc['Subway_Index'] = subway_idx
data_Acc['Bus_Index'] = bus_idx

interval = pd.IntervalIndex.from_tuples([(0,100),(100, 200), (200, 300), (300, 400), (400, 500), (500, float('inf'))])
data_Acc['Subway_Distance_Category'] = pd.cut(data_Acc['Subway_Distance'], bins=interval)
data_Acc['Bus_stop_Distance_Category'] = pd.cut(data_Acc['Bus_Distance'], interval)

plt.figure(figsize=(20,8))
plt.subplot(1,2,1)
sns.barplot(x='Subway_Distance_Category', y='NUMBER OF PERSONS INJURED', data=data_Acc, color='lightgreen')
plt.title('Number of Injuries by Distance to Subway Station', fontsize=16)
plt.xlabel('Distance to Subway Station (meters)', fontsize=14)
plt.ylabel('Number of Injuries', fontsize=14)
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.barplot(x='Bus_stop_Distance_Category', y='NUMBER OF PERSONS INJURED', data=data_Acc, color='lightgreen')
plt.title('Number of Injuries by Distance to Bus Stop', fontsize=16)
plt.xlabel('Distance to Bus Stop (meters)', fontsize=14)
plt.ylabel('Number of Injuries', fontsize=14)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
````
